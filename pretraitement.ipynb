Il faut convertir la taille dans la même unité en k 

```python
def convert_size(size_str):
    if size_str == "Varies with device":
        return None
        
    elif size_str.endswith("M"):
        try:
            number = float(size_str[:-1])
            return number * 1000
        except ValueError:
            return size_str
   
    elif size_str.endswith("k"):
        try:
            return float(size_str[:-1])
        except ValueError:
            return size_str

    else:
        return size_str

data["Size"] = data["Size"].apply(convert_size)

print(data["Size"].unique())

```

```python
#Sur les données != de Varies with device"
stats_size = data["Size"].describe()
print(stats_size)
```

Pour combler les données manquantes, je vais : 
- calculer la médiane du rating selon les catégories
- ajouter cette valeur aux lignes manquantes selon les catégories 

```python
rating_median_by_category = data.groupby("Category")["Rating"].median()
print(rating_median_by_category )
```

```python
data["Rating"] = data.apply(
    lambda row: rating_median_by_category[row["Category"]] if pd.isna(row["Rating"]) else row["Rating"],
    axis=1
)
```
